#### From the Fitbit data, we want to visually understand:

How the average number of steps change by month. Use the appropriate visualization to show the median steps by month. Is Fitbitter more active on weekend or workdays? All plots must be in the same jupyter notebook cell.

**Hints**:

* Use Months_encoded and Week_or Weekend columns.
* Use matplolib.pyplot object oriented API.
* Set your size figure to 12,4
* Explore plt.sca
* Explore plt.xticks
* Save your figures in a folder called `figures` in your repo. 

```python
# Calcular la mediana de pasos por mes
median_steps_per_month = fitbit.groupby('Months_encoded')['Steps'].median()

# Calcular la mediana de pasos por día de la semana 
median_steps_per_day = fitbit.groupby('Work_or_Weekend')['Steps'].median()

# Crear subplots
fig, axs = plt.subplots(nrows=1, ncols=2, figsize=(12, 4))

# Primer subplot: Mediana de pasos por mes
axs[0].bar(median_steps_per_month.index, median_steps_per_month.values, color='skyblue')
axs[0].set_title('Mediana de Pasos por Mes')
axs[0].set_xlabel('Mes')
axs[0].set_ylabel('Mediana de Pasos')
axs[0].set_xticks(median_steps_per_month.index)
axs[0].set_xticklabels(['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic'])

# Segundo subplot: Mediana de pasos por día de la semana
axs[1].bar(median_steps_per_day.index, median_steps_per_day.values, color='lightgreen')
axs[1].set_title('Mediana de Pasos: Trabajo vs Fin de Semana')
axs[1].set_xlabel('Día')
axs[1].set_ylabel('Mediana de Pasos')
axs[1].set_xticks([0, 1])
axs[1].set_xticklabels(['Trabajo', 'Fin de Semana'])

# Ajustar el diseño para evitar solapamientos
plt.tight_layout()

# Guardar la figura
plt.savefig('../figures/fitbit_analysis.png', transparent=False, dpi=80, bbox_inches="tight", format='png')

# Mostrar los gráficos
plt.show()
```

#### Plot the correlation matrix using seaborn.

```python

# Seleccionar solo las columnas numéricas
numeric_cols = titanic.select_dtypes(include=['float64', 'int64'])

# Calcular la matriz de correlación
correlation_matrix = numeric_cols.corr()

# Crear una figura con tamaño (12, 12)
plt.figure(figsize=(12, 12))

# Crear un gráfico de correlación circular
ax = plt.subplot(111, polar=True)
c = ax.matshow(correlation_matrix, cmap='coolwarm')

# Añadir etiquetas
plt.xticks(range(len(correlation_matrix.columns)), correlation_matrix.columns, rotation=90)
plt.yticks(range(len(correlation_matrix.columns)), correlation_matrix.columns)

# Añadir barra de color
plt.colorbar(c)

# Añadir título
plt.title("Gráfico de Correlación Circular del Conjunto de Datos del Titanic", pad=20)

# Mostrar el gráfico
plt.show()
```
