#### Retrieve an arbitary Wikipedia page of "Python" and create a list of links on that page.

````python
# This is the url you will scrape in this exercise
url_5 ='https://en.wikipedia.org/wiki/Python' 
````

````python
response_5 = requests.get(url_5) 
response_5
````

````python
html_5 = response_5.content 
soup_5 = BeautifulSoup(html_5,"html.parser")
soup_5 #elemento parseado 
````

````python
links = soup_5.find_all("a", href=True)#buscamos en las etiquetas a solos los que contengan  href que normalmente tienen los links 
links
````

````python
len(links) #vemos el numero de elementos 
````

````python
link_urls = []    # iniciamos una lista 
for link in links: # para los elementos en link :
    href = link["href"]  #declaramos que el href es un eslice de la key "href"
    if href.startswith("http"): #declaramos que si la la href empieza por "http" lo agregue a la lista esto.
        link_urls.append(href)   #nos da la urls completas 
    
link_urls        
````
