```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import date
#import scikit-learn as sklearn
```

```python
df_iris =pd.read_csv('iris-data.csv')
df_iris
```

```python
df_iris.info()
```

```python
df_iris.describe()
```

```python
#al ser pocos nulos los eliminamos
df_iris = df_iris.dropna()
df_iris
```

```python
# Buscar los valorees unicos en la variable categórica
print(df_iris['class'].unique())
```

```python
print(df_iris['class'].value_counts())
```

```python
#nos quedamos las numericas
numeric = df_iris[["sepal_length_cm", "sepal_width_cm", "petal_length_cm", "petal_width_cm"]]
numeric
```

```python
# Crear un gráfico de caja para cada columna numérica
numeric.plot(kind='box', subplots=True, layout=(2, 2), figsize=(15, 10))
plt.show()
```

```python
# Calcular la matriz de correlación
corr_matrix = numeric.corr()

# Crear un mapa de calor de la matriz de correlación
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

```python
# Eliminar columnas correlacionadas
df_clean=df_iris.drop(columns="petal_length_cm")
df_clean
```

```python
# Eliminar columnas correlacionadas
df_clean1=df_clean.drop(columns="class")
df_clean1
```

```python
# Calcular la matriz de correlación
corr_matrix2 = df_clean1.corr()

# Crear un mapa de calor de la matriz de correlación
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix2, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

```python
#Final data in .csv
df_clean.to_csv('clean_iris-data.csv', index=False, encoding='utf-8')
df_clean
```
